## What is OrientedDet?

**OrientedDet** (`oriented-det`) is an open-source **PyTorch** framework for **oriented and rotated object detection** in aerial and satellite imagery. It focuses on clean geometry, reliable operators, simple datasets, and practical baseline models — without the complexity of large detection mega-frameworks.

![Oriented detections on diagonally parked buses](https://raw.githubusercontent.com/DL4EO/oriented-det/main/demo/out/demo_detections.png)

*Oriented boxes follow vehicle heading instead of swallowing neighbours in axis-aligned rectangles (`demo/demo.jpg`).*

Most detectors describe an object with an **axis-aligned** box. From above there is no “up”: ships, planes, and trucks point every which way, so axis-aligned boxes overlap and waste pixels on background. An **oriented** box adds a rotation (`cx, cy, w, h, θ`) so the box hugs the object and you get heading, length, and width for free.

### What the library covers

- **Geometry** — rotated boxes (`RBox`), quadrilaterals, angle conventions (`le90`), transforms, visualization
- **IoU & NMS** — rotated IoU and oriented NMS (CPU; optional GPU kernels)
- **Datasets** — DOTA, HRSC2016, FAIR1M, SSDD, HRSID loaders; tiling; oriented mAP
- **Models** — Oriented R-CNN, Rotated Faster R-CNN, RetinaNet, FCOS, **RTMDet**, **YOLO-OBB** (ResNet-FPN / CSPNeXt / C2f)
- **Training** — JSON recipes and the **`odet`** CLI (AMP, checkpointing, TensorBoard)
- **Export & Hub** — ONNX export (`odet export`); pretrained weights via Hugging Face Hub

This tutorial teaches the library end to end. For the full repository map, citation, and install variants, see the [GitHub README](https://github.com/DL4EO/oriented-det#orienteddet) or [tutorial/README.md](README.md).
